# GUI widgets
This notebook installs `ipywidgets` into the browser kernel and builds an interactive form. The first installation requires an internet connection.

```python
import ipywidgets as widgets
from IPython.display import display

quantity = widgets.IntSlider(value=1, min=1, max=20, description="Quantity")
category = widgets.Dropdown(options=["books", "music", "software"], description="Category")
notes = widgets.Textarea(description="Notes")
submit = widgets.Button(description="Submit", button_style="primary")
output = widgets.Output()

def handle_submit(button):
    with output:
        output.clear_output()
        print(f"Submitted {quantity.value} {category.value} item(s). {notes.value}")

submit.on_click(handle_submit)
display(widgets.VBox([quantity, category, notes, submit, output]))
```
